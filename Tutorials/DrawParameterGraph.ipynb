## Draw essential parameter graph

```python
net_spec = 'x : ~y x : E\n y : ~x y : E'

network = DSGRN.Network(net_spec)

DSGRN.DrawGraph(network)
```

```python
parameter_graph = DSGRN.ParameterGraph(network)

print('Number of parameter nodes:', parameter_graph.size())
```

```python
DSGRN.draw_parameter_graph(parameter_graph)
```

Draw Boolean parameter nodes as red squares and remaining parameters are green circles.

```python
bool_params = DSGRN.boolean_parameters(parameter_graph)

node_shapes = {v: 'square' if v in bool_params else 'circle' for v in range(parameter_graph.size())}
node_colors = {v: 'red' if v in bool_params else 'lightgreen' for v in range(parameter_graph.size())}

DSGRN.draw_parameter_graph(parameter_graph, node_shapes=node_shapes, node_colors=node_colors)
```

Classify nodes according to the number of stable Morse nodes and color each stability class with a different color.

```python
node_classes = DSGRN.Blowup.StabilityQuery(network)

DSGRN.draw_parameter_graph(parameter_graph, node_classes=node_classes)
```

Color nodes according to their stability and plot Boolean nodes as squares.

```python
DSGRN.draw_parameter_graph(parameter_graph, node_classes=node_classes, node_shapes=node_shapes)
```

## Use `networkx` to draw parameter graph

```python
DSGRN.draw_parameter_graph_nx(parameter_graph, node_classes=node_classes, fig_w=20, fig_h=20)
```

## Choose nodes layout

```python
DSGRN.draw_parameter_graph_nx(parameter_graph, node_classes=node_classes, layout='grid', fig_w=15, fig_h=15)
```
